```python
import numpy as np
import matplotlib.pyplot as plt

# Define the x range
x = np.linspace(-3, 3, 400)

# Define the functions
p = np.exp(-x**2 / 2)
h = -np.log(p)  # Equivalent to x**2 / 2

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(x, p, label=r'$p(x_1) = \exp(-x_1^2 / 2)$', color='blue')
plt.plot(x, h, label=r'$s(x_1) = -\log(p(x_1))$', color='red')
plt.title('Plot of $p(x_1)$ and $s(x_1)$')
plt.xlabel('$x_1$')
plt.ylabel('Function values')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid setup
x = np.linspace(-4, 4, 200)
y = np.linspace(-4, 4, 200)
X, Y = np.meshgrid(x, y)

# Define the 2D Gaussian (mean 0, covariance identity)
Z_p = np.exp(-(X**2 + Y**2) / 2) / (2 * np.pi)  # p(x1, x2)
Z_h = -np.log(Z_p)  # h(x1, x2) = -log(p)

# Plotting
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# p(x1, x2)
cp1 = axs[0].contourf(X, Y, Z_p, levels=50, cmap='viridis')
axs[0].set_title(r'$p(x_1, x_2)$')
axs[0].set_xlabel('$x_1$')
axs[0].set_ylabel('$x_2$')
fig.colorbar(cp1, ax=axs[0])

# h(x1, x2)
cp2 = axs[1].contourf(X, Y, Z_h, levels=50, cmap='inferno')
axs[1].set_title(r'$h(x_1, x_2) = -\log(p(x_1, x_2))$')
axs[1].set_xlabel('$x_1$')
axs[1].set_ylabel('$x_2$')
fig.colorbar(cp2, ax=axs[1])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Grid
x = np.linspace(-4, 4, 100)
y = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(x, y)

# Probability density (Z_p) and negative log (Z_h)
Z_p = (1 / (2 * np.pi)) * np.exp(-(X**2 + Y**2) / 2)
Z_h = -np.log(Z_p)  # This becomes (X^2 + Y^2)/2 + const

# Plotting
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot Z_p (Gaussian hill)
surf_p = ax.plot_surface(X, Y, Z_p, cmap='viridis', alpha=0.9, edgecolor='none', label='p(x)')
# Plot Z_h (Negative log, bowl-shaped)
surf_h = ax.plot_surface(X, Y, Z_h / 50, cmap='inferno', alpha=0.6, edgecolor='none', label='-log(p(x))')
# Note: We divide Z_h by 50 to fit both surfaces in the same z-axis range

# Labels and title
ax.set_title('3D Plot of $p(x_1, x_2)$ and $h(x_1, x_2) = -\log(p)$')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('Value')

# Optional: Add manual legend
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='blue', lw=4, label=r'$p(x_1, x_2)$'),
    Line2D([0], [0], color='orange', lw=4, label=r'$h(x_1, x_2) = -\log(p)$ (rescaled)')
]
ax.legend(handles=legend_elements)

plt.tight_layout()
plt.show()
```
